### Tuples

A tuple is:
- Ordered — elements maintain their positions.
- Immutable — you cannot modify its elements after creation.
- Allows duplicates.
- Allows different data types.

Remember this rule:
 - In Python, the comma creates a tuple, not the parentheses.

### Why do we need tuples?
Imagine an AI application storing a model's configuration:
- model_config = ("gemini-2.5-flash", 0.7, 2048)

- You don't want these configuration values accidentally modified throughout your application.
- A tuple communicates that these values should remain fixed.

```python
a = (10) # <class 'int'>
b = (10,) # <class 'tuple'>
c = 10, # <class 'tuple'>

#Again, the comma creates the tuple. Parentheses are optional here.
```

#### But remember: immutability doesn't mean every object referenced by a tuple is immutable. We'll explore that later.

```python
data = (10, 20, [30, 40])
data[2].append(50)
print(data) # (10, 20, [30, 40, 50])
```

```python
x = (10, 20)
y = 10, 20

print(type(x) == type(y)) # True
print(x == y) # True
```

### Tuple unpacking
This is particularly useful in real-world Python development.

```python
# Python automatically assigns each tuple element to the corresponding variable.

student = ("Alice", 22, "AI Engineering")

name, age, course = student

print(name) # Alice
print(age) # 22
print(course) # AI Enigneering
```

```python
coordinates = (10, 20)

x, y = coordinates

x, y = y, x

print(x, y) # 20 10
```

### Extended unpacking with *
- The * operator collects the remaining elements into a list.
- You can use only one starred variable in an unpacking assignment.

```python
numbers = (10, 20, 30, 40, 50, 60)

first, *middle, last = numbers

print(first) # 10
print(middle) # [20, 30, 40] 
print(last) # 50
```

```python
data = (5, 10, 15, 20, 25, 30)

first, *remaining = data

print(first) # 5
print(remaining) # [ 10, 15, 20, 25, 30] 
```

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

middle.append(60)

print(first) # 10
print(middle) # [20, 30, 40, 60]
print(last) # 50 
print(numbers) # (10, 20, 30, 40, 50)
```

### Tuple methods
Tuples have only two built-in methods:
  - count(value) -- Counts occurrences
  - index(value) -- Returns the first matching index
  - index(value, start) -- searches from the specified starting index.

```python
numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))  # 3
print(numbers.index(20)) # 1
```

```python
data = (5, 10, 20, 10, 30, 10)

print(data.count(10)) # 3
print(data.index(10)) # 1
print(data.index(10, 2)) # 3
```

### Tuple slicing

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4]) # (20, 30, 40)
print(numbers[:3]) # (10, 20, 30)
print(numbers[::2]) # (10, 30, 50)
print(numbers[::-1]) # (60, 50, 40, 30, 20, 10)
```

### Nested tuples

```python
students = (
    ("Alice", 85),
    ("Bob", 90),
    ("Charlie", 78)
)

print(students[0])     # ("Alice", 85)
print(students[0][0])  # Alice
print(students[1][1])  # 90
```

```python
data = (
    ("Python", 90),
    ("Java", 80),
    ("AI", 95)
)

print(data[2][0]) # AI
print(data[0][1]) # 90
print(data[-1][-1]) # 95
```

### Tuple packing and unpacking in real-world Python

Important: The * operator has different roles depending on where it appears.
- Function definition: collects arguments.
- Function call: unpacks arguments.

```python
#unpacking
response = ("success", "Generated successfully", 200)
status, message, status_code = response

#packing
status = "success"
message = "Generated successfully"
status_code = 200

response = status, message, status_code
```

```python
def get_user():
    return "Alice", 25, "AI Engineer"

name, age, role = get_user()

print(name)
print(age)
print(role)
```

### Unpacking with *args and tuples

```python
def calculate_total(a, b, c):
    return a + b + c

numbers = (10, 20, 30)

print(calculate_total(*numbers))
```

```python
def show_details(*args):
    print(type(args)) # <class 'tuple'>
    print(args) # (10, 20, 30, 40)
    print(len(args)) # 4

data = (10, 20, 30, 40)

show_details(*data)
```

### Final challenge

```python
def process_data(*args):
    first, *middle, last = args

    return first, middle, last


result = process_data(10, 20, 30, 40, 50)

print(result) # (10, [20, 30, 40], 50)
print(type(result)) # <class 'tuple'>
print(type(result[1])) # <class 'list'>
```
