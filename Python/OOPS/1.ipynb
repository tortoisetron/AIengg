### Class and Object

```python
# class create the blueprint for the object. It defines the properties and behaviors of the object.
# The class is a template for creating objects. The class defines the attributes and methods of the object.
# The attributes are the properties of the object and the methods are the behaviors of the object.

class Car: 
    def __init__(self, make, model, year): 
        self.make = make 
        self.model = model 
        self.year = year 

    def display_info(self): 
        print(f"Car: {self.year} {self.make} {self.model}")

car1 = Car("Toyota", "Camry", 2020)
car2 = Car("Honda", "Civic", 2019)
car2.display_info()  # Output: Car: 2020 Toyota Camry
```

### Dunder method

```python
# Dunder methods are special methods that have double underscores at the beginning and end of their names.
# They are also called magic methods or special methods. Dunder methods are used to define the behavior of objects in Python.
# They are used to implement operator overloading, object representation, and other special behaviors.

#__init__ is a dunder method that is called when an object is created. It is used to initialize the attributes of the object.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person: {self.name}, Age: {self.age}"
    
person1 = Person("Alice", 30)
print(person1)  # Output: Person: Alice, Age: 30

#__add__ is a dunder method that is called when the + operator is used with objects. It is used to define the behavior of the + operator for objects.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other): # other is the object that is being added to the current object. It is passed as an argument to the __add__ method.
        return self.x + other.x, self.y + other.y

vector1 = Vector(1, 2)
vector2 = Vector(3, 4)
vector3 = vector1 + vector2
print(vector3)  # Output: Vector: (4, 6)

# __len__ is a dunder method that is called when the len() function is used with objects. 
# It is used to define the behavior of the len() function for objects.

class MyList:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)
    
my_list = MyList([1, 2, 3, 4, 5])
print(len(my_list))  # Output: 5

# __getitem__ is a dunder method that is called when the indexing operator [] is used with objects.
# It is used to define the behavior of the indexing operator for objects.

class MyList:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index]
    
my_list = MyList([1, 2, 3, 4, 5])
print(my_list[2])  # Output: 3

# __setitem__ is a dunder method that is called when the indexing operator [] is used with objects to set a value.
# __getitem__ is a dunder method that is called when the indexing operator [] is used with objects to get a value.

class MyList:
    def __init__(self, items):
        self.items = items

    def __setitem__(self, index, value):
        self.items[index] = value
    
    # withoout this error will be thrown when trying to access the value at the index.
    def __getitem__(self, index):
        return self.items[index]

my_list = MyList([1, 2, 3, 4, 5])
my_list[2] = 10
print(my_list[2])  # Output: 10

# __iter__ is a dunder method that is called when the iter() function is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __iter__(self):
        return iter(self.items)

my_list = MyList([1, 2, 3, 4, 5])
for item in my_list:
    print(item, end="")  # Output: 1 2 3 4 5 
    # end=" " is used to print the items in the same line with a space in between. Without it, the items will be printed in separate lines.

# __contains__ is a dunder method that is called when the in operator is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __contains__(self, item):
        return item in self.items
    
my_list = MyList([1, 2, 3, 4, 5])
print(3 in my_list)  # Output: True

# __call__ is a dunder method that is called when the object is called as a function.
# It is used to define the behavior of the object when it is called as a function.
class MyList:
    def __init__(self, items):
        self.items = items

    def __call__(self):
        return self.items

my_list = MyList([1, 2, 3, 4, 5])
print(my_list())  # Output: [1, 2, 3, 4, 5]

# __del__ is a dunder method that is called when the del operator is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __del__(self):
        print("Deleting MyList object")

my_list = MyList([1, 2, 3, 4, 5])
del my_list  # Output: Deleting MyList object
```

### Decorator

```python
# decorators are a way to modify the behavior of a function or a class. 
# They are used to add functionality to a function or a class without modifying its source code.

def MyList(func):
    # *args and **kwargs are used to pass a variable number of arguments to the function.
    # *args is used to pass a variable number of positional arguments to the function.
    # **kwargs is used to pass a variable number of keyword arguments to the function.
    
    def wrapper(*args, **kwargs): 
        print("Before calling the function")
        result = func(*args, **kwargs)
        print("After calling the function")
        return result
    return wrapper

@MyList
def my_function(x, y):
    return x + y

my_function = my_function(2,3)
print(my_function)  # Output: 5

# decoraator with try and finally block to handle exceptions and ensure that the function is called even if an exception is raised.
def MyList(func):
    def wrapper(*args, **kwargs): 
        try:
            print("Before calling the function")
            result = func(*args, **kwargs)
            return result 
        except Exception as e:
            print(f"An exception occurred: {e}")
        finally:
            print("After calling the function")
    return wrapper

@MyList
def my_function(x, y):
    # raise ValueError("An error occurred")  # This will raise an exception and the finally block will be executed.
    return x + y

print(my_function(2,3)) # Output: Before calling the function
                        #         After calling the function -- finally block execute first while return id being executed
                        #         5
                        
# functiontools.wraps is a decorator that is used to preserve the metadata of the original function when it is decorated.
from functools import wraps

def MyList(func):
    @wraps(func)
    def wrapper(*args, **kwargs): 
        print("Before calling the function")
        result = func(*args, **kwargs)
        print("After calling the function")
        return result
    return wrapper


@MyList
def my_function(x, y):
    return x + y

print(my_function.__name__)  # Output: my_function
                        
```

### Instance method, Class method and Static method

```python
# An instance method is a function inside a class that operates on a specific object (instance) rather than the class as a whole.
from unittest import result


class MyClass:
    def __init__(self, value):
        self.value = value

    def instance_method(self):
        return f"Instance method called. Value: {self.value}"

print(MyClass(10).instance_method())  # Output: Instance method called. Value: 10

#-----------------------------------------------------------------------------------------------------------------------------------#

# A class method is a function inside a class that operates on the class itself rather than on instances of the class.
# It is defined using the @classmethod decorator and takes the class as its first argument, conventionally named cls.
class MyClass:
    class_variable = "I am a class variable"

    @classmethod
    def class_method(cls):
        return f"Class method called. Class variable: {cls.class_variable}"

my_class = MyClass
print(my_class.class_method())  # Output: Class method called. Class variable: I am a class variable
print(my_class.class_variable)  # Output: I am a class variable

# what is the use of class method?
# Class methods are used for operations that are related to the class itself rather than any specific instance

class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    @classmethod
    def from_dict(cls, data):
        return cls(
            name = data['name'],
            age = data['age']
        )
        
student1 = Student('Alice',50)
print(student1.name)
data = {
    'name' : 'Tortoise',
    'age' : 25
}
student2 = Student.from_dict(data)
print(student2.name)

#-----------------------------------------------------------------------------------------------------------------------------------#

# A @staticmethod in Python is a method bound to a class rather than an object instance, 
# meaning it does not receive an implicit first argument like self (instance) or cls (class)

class Math:
    @staticmethod
    def add(a,b):
        return a + b
    
result = Math.add(5,6)
print(result)
```

### Getter and Setter

```python
# Getter: The getter method is used to retrieve the value of a private attribute. It allows controlled access to the attribute.
# Setter: The setter method is used to set or modify the value of a private attribute. 
# It allows you to control how the value is updated, enabling validation or modification of the data before it’s actually assigned.

class Geek: 
	def __init__(self, age = 0): 
		self._age = age 
	
	# getter method 
	def get_age(self): 
		return self._age 
	
	# setter method 
	def set_age(self, x): 
		self._age = x 

raj = Geek(26) 
# retrieving age using getter 
print(raj.get_age()) 

# setting the age using setter 
raj.set_age(21) 

# retrieving age using getter 
print(raj.get_age()) 

print(raj._age)

#-----------------------------------------------------------------------------------------------------------------------------------#
# Using @property decorators
class Geeks: 
	def __init__(self): 
		self._age = 0
	
	# using property decorator 
	# a getter function 
	@property
	def age(self): 
		print("getter method called") 
		return self._age 
	
	# a setter function 
	@age.setter 
	def age(self, a): 
		if(a < 18): 
			raise ValueError("Sorry you age is below eligibility criteria") 
		print("setter method called") 
		self._age = a 

#Instantiation object
mark = Geeks() 
#setting value
mark.age = 19
#getting the value
print(mark.age)
```

### Inheritance

```python
# Inheritance is a way to create a new class that is a modified version of an existing class.
# The new class is called the child class and the existing class is called the parent/base class.

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")
        # raise NotImplementedError("Subclasses must implement this method")
        # pass

class Dog(Animal):
    # method overriding is a feature of object-oriented programming that allows a subclass to provide a 
    # specific implementation of a method that is already defined in its superclass.
    def speak(self):
        return "Woof!"
    
class Cat(Animal):
    def speak(self):
        # calling the parent class method to show how to call the parent class method from the child class method.
        super().speak()
        print(f"{self.name} says: Meow!")
    
dog = Dog("Buddy")
cat = Cat("Whiskers")
cat.speak()  # Output: # Whiskers makes a sound.
                    #Whiskers says: Meow!
```

### Multiple Inheritance & MRO


```python
# Multiple inheritance in Python allows a child class to inherit attributes and methods from more than one parent class.

class Car:
    def engine(self):
        print('Engine start')

class ElectricCar:
    def battery(self):
        print('Engine start silent')
        
class Vehicles(Car,ElectricCar):
    pass

bike = Vehicles()
bike.battery()
bike.engine()


# MRO stands for Method Resolution Order. It determines the order Python searches classes when looking for a method.
class Camera:
    def start(self):
        print("Camera started")
        super().start()

class Phone:
    def start(self):
        print("Phone started")


class Smartphone(Camera, Phone):
    def start(self):
        print("Smartphone started")
        super().start()

# Smartphone → Camera → Phone → object
device = Smartphone()
device.start()
```

### Polymorphism and Duck Typing

```python
# Polymorphism in Python refers to the ability of different objects, functions, or operators to share the same interface (name)
# but behave differently based on the data type or context.

class CreditCard:
    def process_payment(self, amount):
        return f"Charging ${amount} to Credit Card."

class PayPal:
    def process_payment(self, amount):
        return f"Routing ${amount} through PayPal."

# This function exhibits polymorphic behavior without needing inheritance
def checkout(payment_gateway, amount):
    print(payment_gateway.process_payment(amount))

checkout(CreditCard(), 100)
checkout(PayPal(), 50)

#-----------------------------------------------------------------------------------------------------------------------------------#

# Duck Typing is a programming concept where an object's suitability is determined by the presence of specific methods 
# and properties rather than its actual class or explicit type declaration.

# Python program to demonstrate
# duck typing

class Bird:
    def fly(self):
        print("fly with wings")

class Airplane:
    def fly(self):
        print("fly with fuel")

class Fish:
    def swim(self):
        print("fish swim in sea")

# Attributes having same name are
# considered as duck typing
for obj in Bird(), Airplane(), Fish():
    obj.fly() # Output: fly with wings
            #fly with fuel
            # AttributeError cause Fish class doesn't have fly() so Duck-typing emphasis what the object can really do, 
            # rather than what the object is.
```

### Abstract class

```python
# An abstract class is a restricted class that cannot be instantiated directly to create an object. 
# Instead, it serves as a blueprint or template for other classes.
# the process of hiding implementation complexities while exposing essential behaviors.

from abc import ABC, abstractmethod

# 1. Define the Abstract Class
class Vehicle(ABC):
    
    # An abstract method (must be implemented by subclasses)
    @abstractmethod
    def start_engine(self):
        pass

    # A concrete method (inherited as-is by subclasses)
    def honk(self):
        return "Beep beep!"

# 2. Implement a Subclass
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started. Vroom!"

# --- Execution ---

# Trying to instantiate the abstract class will fail:
# my_vehicle = Vehicle()  # Raises TypeError

# Instantiating the concrete subclass works perfectly:
my_car = Car()
print(my_car.start_engine())  # Output: Car engine started. Vroom!
print(my_car.honk())          # Output: Beep beep!
```

### Encapsulation

```python
# Encaptulation is bout keeping the data and the operation that manage it together, while controlling how that data is accessed and modified.
#protecting data inside a class
# _single-underscore -- protected
# __double-underscore -- private (_name__mangling)

class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self._gender = gender # Property property
        self.__age = age # Private property
        
    # Use a getter method to access a private property:
    def get_age(self):
            return self.__age

p1 = Person("Emil", 25, 'Male')
p1._gender = "Female"
p1.__age = 23 # this works cause it create a new object, not using the class attribute 
p1._Person__age = 24 # this directly change the class attribute
print(p1.name)
print(p1._gender)
print(p1._Person__age)
# print(p1.__age) #if create new object then it works else This will cause an error
print(p1.get_age())
```

### Composition vs Inheritance

```python
# Composition - A class 'has-a' component (contains an instance of another class).
# Inheritance - A subclass 'is-a' specialized version of a base class.

class BaseClass:
    def __init__(self,name,age):
        self.name = name
        self.age = age

#Composition
class Composition:
    def __init__(self,name,age):
        self.base = BaseClass(name,age)
    @property
    def get(self):
        print(self.base.name)
        print(self.base.age)

#Inheritance
class SubClass(BaseClass):
    def __init__(self,name,age):
        super().__init__(name,age)
        self.name = name
        
    @property
    def get(self):
        print(self.name)
        print(self.age)
        
obj = SubClass('test',12).get
obj = Composition('test',12).get
```
