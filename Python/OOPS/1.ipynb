### Class and Object

```python
# class create the blueprint for the object. It defines the properties and behaviors of the object.
# The class is a template for creating objects. The class defines the attributes and methods of the object.
# The attributes are the properties of the object and the methods are the behaviors of the object.

class Car: 
    def __init__(self, make, model, year): 
        self.make = make 
        self.model = model 
        self.year = year 

    def display_info(self): 
        print(f"Car: {self.year} {self.make} {self.model}")

car1 = Car("Toyota", "Camry", 2020)
car2 = Car("Honda", "Civic", 2019)
car2.display_info()  # Output: Car: 2020 Toyota Camry
```

### Dunder method

```python
# Dunder methods are special methods that have double underscores at the beginning and end of their names.
# They are also called magic methods or special methods. Dunder methods are used to define the behavior of objects in Python.
# They are used to implement operator overloading, object representation, and other special behaviors.

#__init__ is a dunder method that is called when an object is created. It is used to initialize the attributes of the object.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person: {self.name}, Age: {self.age}"
    
person1 = Person("Alice", 30)
print(person1)  # Output: Person: Alice, Age: 30

#__add__ is a dunder method that is called when the + operator is used with objects. It is used to define the behavior of the + operator for objects.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other): # other is the object that is being added to the current object. It is passed as an argument to the __add__ method.
        return self.x + other.x, self.y + other.y

vector1 = Vector(1, 2)
vector2 = Vector(3, 4)
vector3 = vector1 + vector2
print(vector3)  # Output: Vector: (4, 6)

# __len__ is a dunder method that is called when the len() function is used with objects. 
# It is used to define the behavior of the len() function for objects.

class MyList:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)
    
my_list = MyList([1, 2, 3, 4, 5])
print(len(my_list))  # Output: 5

# __getitem__ is a dunder method that is called when the indexing operator [] is used with objects.
# It is used to define the behavior of the indexing operator for objects.

class MyList:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index]
    
my_list = MyList([1, 2, 3, 4, 5])
print(my_list[2])  # Output: 3

# __setitem__ is a dunder method that is called when the indexing operator [] is used with objects to set a value.
# __getitem__ is a dunder method that is called when the indexing operator [] is used with objects to get a value.

class MyList:
    def __init__(self, items):
        self.items = items

    def __setitem__(self, index, value):
        self.items[index] = value
    
    # withoout this error will be thrown when trying to access the value at the index.
    def __getitem__(self, index):
        return self.items[index]

my_list = MyList([1, 2, 3, 4, 5])
my_list[2] = 10
print(my_list[2])  # Output: 10

# __iter__ is a dunder method that is called when the iter() function is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __iter__(self):
        return iter(self.items)

my_list = MyList([1, 2, 3, 4, 5])
for item in my_list:
    print(item, end="")  # Output: 1 2 3 4 5 
    # end=" " is used to print the items in the same line with a space in between. Without it, the items will be printed in separate lines.

# __contains__ is a dunder method that is called when the in operator is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __contains__(self, item):
        return item in self.items
    
my_list = MyList([1, 2, 3, 4, 5])
print(3 in my_list)  # Output: True

# __call__ is a dunder method that is called when the object is called as a function.
# It is used to define the behavior of the object when it is called as a function.
class MyList:
    def __init__(self, items):
        self.items = items

    def __call__(self):
        return self.items

my_list = MyList([1, 2, 3, 4, 5])
print(my_list())  # Output: [1, 2, 3, 4, 5]

# __del__ is a dunder method that is called when the del operator is used with objects.
class MyList:
    def __init__(self, items):
        self.items = items

    def __del__(self):
        print("Deleting MyList object")

my_list = MyList([1, 2, 3, 4, 5])
del my_list  # Output: Deleting MyList object
```

### Decorator

```python
# decorators are a way to modify the behavior of a function or a class. 
# They are used to add functionality to a function or a class without modifying its source code.

def MyList(func):
    # *args and **kwargs are used to pass a variable number of arguments to the function.
    # *args is used to pass a variable number of positional arguments to the function.
    # **kwargs is used to pass a variable number of keyword arguments to the function.
    
    def wrapper(*args, **kwargs): 
        print("Before calling the function")
        result = func(*args, **kwargs)
        print("After calling the function")
        return result
    return wrapper

@MyList
def my_function(x, y):
    return x + y

my_function = my_function(2,3)
print(my_function)  # Output: 5

# decoraator with try and finally block to handle exceptions and ensure that the function is called even if an exception is raised.
def MyList(func):
    def wrapper(*args, **kwargs): 
        try:
            print("Before calling the function")
            result = func(*args, **kwargs)
            return result 
        except Exception as e:
            print(f"An exception occurred: {e}")
        finally:
            print("After calling the function")
    return wrapper

@MyList
def my_function(x, y):
    # raise ValueError("An error occurred")  # This will raise an exception and the finally block will be executed.
    return x + y

print(my_function(2,3)) # Output: Before calling the function
                        #         After calling the function -- finally block execute first while return id being executed
                        #         5
                        
# functiontools.wraps is a decorator that is used to preserve the metadata of the original function when it is decorated.
from functools import wraps

def MyList(func):
    @wraps(func)
    def wrapper(*args, **kwargs): 
        print("Before calling the function")
        result = func(*args, **kwargs)
        print("After calling the function")
        return result
    return wrapper


@MyList
def my_function(x, y):
    return x + y

print(my_function.__name__)  # Output: my_function
                        
```

### Inheritance

```python
# Inheritance is a way to create a new class that is a modified version of an existing class.
# The new class is called the child class and the existing class is called the parent/base class.

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")
        # raise NotImplementedError("Subclasses must implement this method")
        # pass

class Dog(Animal):
    # method overriding is a feature of object-oriented programming that allows a subclass to provide a 
    # specific implementation of a method that is already defined in its superclass.
    def speak(self):
        return "Woof!"
    
class Cat(Animal):
    def speak(self):
        # calling the parent class method to show how to call the parent class method from the child class method.
        super().speak()
        print(f"{self.name} says: Meow!")
    
dog = Dog("Buddy")
cat = Cat("Whiskers")
cat.speak()  # Output: # Whiskers makes a sound.
                    #Whiskers says: Meow!
```
