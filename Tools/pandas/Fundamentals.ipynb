```python
df = pd.DataFrame({
    "Student": ["Rahul", "Amit", "Priya"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
})
# Greater than
df[df["Marks"] > 80]

# AND
df[(df["Marks"] > 80) & (df["Age"] > 20)]

# OR
df[(df["Marks"] > 80) | (df["Age"] > 21)]

# NOT
df[~(df["Marks"] > 80)]
```

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Rahul", "Amit", "Priya", "Neha", "Rohan"],
    "Age": [21, 22, None, 23, 19],
    "Marks": [85, None, 78, 65, 95]
})

print(df)
df.isna()
df["Marks"].isna()
df["Marks"].isna().sum()
df.isna().sum()
df.isna().sum().sum()
df["Marks"].notna()
df[df["Marks"].notna()]
df[df["Age"].isna()]
df.dropna()
df["Marks"].fillna(0,inplace=True)


```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [20, None, 30, 40],
    "Marks": [80, 90, None, 70]
})

missing_value = df.isna().sum()
filling_age = df['Age'].fillna(df["Age"].median())
filling_marks = df['Marks'].fillna(df["Marks"].mean())
print(missing_value,filling_age,filling_marks)
```
