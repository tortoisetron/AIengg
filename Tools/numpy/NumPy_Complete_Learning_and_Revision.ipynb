## 1. Why NumPy?

A Python list is a general-purpose container. A NumPy array is designed for numerical computation.

With a list, multiplying by 2 repeats the list. With a NumPy array, multiplying by 2 multiplies every number.

NumPy is useful in data analysis, scientific computing, machine learning, image processing, and many AI workflows.

```python
import numpy as np

python_list = [10, 20, 30]
numpy_array = np.array([10, 20, 30])

print(python_list * 2)
print(numpy_array * 2)
```

**Teaching point:** `list * 2` repeats the sequence; `array * 2` performs element-wise numerical multiplication.

**Important:** NumPy is generally faster for large numerical workloads because many operations run in optimized compiled code. It is not guaranteed to be faster for every tiny operation.

## 2. Creating NumPy arrays

The usual way to create an array from existing values is `np.array()`.

```python
a = np.array([10, 20, 30])                 # 1D
b = np.array([[1, 2, 3], [4, 5, 6]])       # 2D
c = np.array([[[1, 2], [3, 4]]])           # 3D

print(a)
print(b)
print(c)
```

### Common array creation functions

- `np.zeros(shape)`: fills an array with zeros.
- `np.ones(shape)`: fills an array with ones.
- `np.full(shape, value)`: fills an array with a chosen value.
- `np.arange(start, stop, step)`: uses a step; the stop value is excluded.
- `np.linspace(start, stop, num)`: creates a chosen number of evenly spaced values; the endpoint is included by default.
- `np.eye(n)`: creates an identity matrix.
- `np.empty(shape)`: allocates an array without initializing its values; its contents should not be assumed to be zero.

```python
print(np.zeros((2, 3)))
print(np.ones((2, 2)))
print(np.full((2, 3), 7))
print(np.arange(0, 10, 2))
print(np.linspace(0, 10, 5))
print(np.eye(3))
```

### `arange()` vs `linspace()`

Use `arange` when the step size matters. Use `linspace` when the number of points matters.

```python
print(np.arange(0, 10, 2))  # 0, 2, 4, 6, 8
print(np.linspace(0, 10, 5))  # 0, 2.5, 5, 7.5, 10
```

## 3. Dimensions and array attributes

An array's dimension count tells you how many axes it has:
- 0D: one scalar value
- 1D: one axis, like a sequence
- 2D: rows and columns, like a table
- 3D: a stack or collection of 2D arrays

```python
zero_d = np.array(5)
one_d = np.array([10, 20, 30])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
three_d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

for arr in (zero_d, one_d, two_d, three_d):
    print("ndim:", arr.ndim, "shape:", arr.shape, "size:", arr.size)
```

Useful attributes:

| Attribute | Meaning |
|---|---|
| `ndim` | Number of dimensions / axes |
| `shape` | Length along each axis |
| `size` | Total number of elements |
| `dtype` | Data type of elements |
| `itemsize` | Bytes used by one element |

For a 2D array with shape `(2, 3)`, there are 2 rows, 3 columns, and 6 total elements.

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])
print(arr.ndim)       # 2
print(arr.shape)      # (2, 3)
print(arr.size)       # 6
print(arr.dtype)
print(arr.itemsize)
```

## 4. Data types and conversion

NumPy arrays usually store elements using one consistent data type. Common types include integers, floating-point numbers, and booleans.

```python
integers = np.array([1, 2, 3])
decimals = np.array([1.5, 2.5, 3.5])
flags = np.array([True, False, True])

print(integers.dtype)
print(decimals.dtype)
print(flags.dtype)
```

You can request a type during creation with `dtype=`, or convert an existing array with `.astype()`.

```python
values = np.array([1, 2, 3])
as_float = values.astype(float)

print(as_float)
print(as_float.dtype)

decimal_values = np.array([1.9, 2.7, 3.2])
print(decimal_values.astype(int))  # conversion truncates the fractional part
```

**Interview note:** `astype(int)` does not round to the nearest integer; it truncates the fractional part for ordinary positive values.

## 5. Indexing and slicing

Indexing selects individual elements. Python/NumPy indexing starts at 0.

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])     # first element
print(arr[-1])    # last element
print(arr[1:4])   # indices 1, 2, 3
print(arr[:3])    # from start up to (not including) index 3
print(arr[::2])   # every second element
```

### Indexing and slicing 2D arrays

Use `array[row, column]`. A colon `:` means select all entries along that axis.

```python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print(matrix[1, 2])   # row 1, column 2 -> 60
print(matrix[0, :])   # all columns from row 0
print(matrix[:, 1])   # all rows from column 1
print(matrix[0:2, 1:3])
```

**Remember:** the stop index in a slice is excluded. `matrix[:, 1]` selects the second column from every row.

## 6. Reshaping and changing dimensions

`reshape()` changes the shape without changing the sequence of values. The total number of elements must remain the same.

```python
values = np.arange(1, 13)
grid = values.reshape(3, 4)

print(values)
print(grid)
print(grid.shape)
```

### Let NumPy infer one dimension with `-1`

You can use `-1` for one dimension and NumPy calculates it from the element count.

```python
values = np.arange(12)
print(values.reshape(3, -1))  # NumPy infers 4 columns
print(values.reshape(-1, 2))  # NumPy infers 6 rows
```

### `flatten()` vs `ravel()`

Both turn a multidimensional array into a 1D array.
- `flatten()` always returns a copy.
- `ravel()` returns a view when possible; otherwise it returns a copy.

```python
matrix = np.array([[1, 2], [3, 4]])

flat_copy = matrix.flatten()
flat_ravel = matrix.ravel()

flat_copy[0] = 99
print("Original after flatten change:", matrix)

flat_ravel[0] = 77
print("Original after ravel change (for this array):", matrix)
```

### Add a dimension with `np.newaxis` / `None`

A 1D array can be shaped like a row or a column by adding an axis.

```python
x = np.array([10, 20, 30])
row = x[np.newaxis, :]
column = x[:, np.newaxis]

print(row, row.shape)
print(column, column.shape)
```

## 7. Copy vs view

A **view** shares the same underlying data with the original array. A **copy** owns separate data. Changing a view can change the original; changing a copy does not.

```python
original = np.array([10, 20, 30, 40])

view = original[1:3]
copy = original[1:3].copy()

view[0] = 999
print("After changing view:", original)

copy[0] = 555
print("After changing copy:", original)
print("Copy:", copy)
print("Shares memory?", np.shares_memory(original, view))
```

**Important:** basic slicing commonly creates views, while advanced indexing (such as integer-array or boolean-array indexing) generally creates copies. When you need independent data, use `.copy()` explicitly.

## 8. Sorting arrays

`np.sort()` returns a sorted copy. The array method `.sort()` sorts the original array in place.

```python
values = np.array([40, 10, 30, 20])
sorted_copy = np.sort(values)

print("Sorted copy:", sorted_copy)
print("Original:", values)

values.sort()
print("Original after .sort():", values)
```

For a 2D array, specify an axis when you want to sort each row or column.

```python
matrix = np.array([[3, 1, 2],
                   [9, 7, 8]])
print(np.sort(matrix, axis=1))  # sort each row
print(np.sort(matrix, axis=0))  # sort each column
```

## 9. Boolean indexing and filtering

A comparison creates a boolean array (a mask). Use that mask to select values that meet a condition.

```python
marks = np.array([35, 48, 67, 82, 55, 91])

passed_mask = marks >= 50
print(passed_mask)
print(marks[passed_mask])
print(marks[marks >= 50])
```

Combine conditions with:
- `&` for AND
- `|` for OR
- `~` for NOT

Put each comparison inside parentheses. Use these operators with NumPy arrays rather than Python's `and` / `or`.

```python
marks = np.array([35, 48, 67, 82, 55, 91])

print(marks[(marks >= 50) & (marks < 80)])
print(marks[(marks < 40) | (marks > 85)])
print(marks[~(marks >= 50)])
```

### Filtering rows in a 2D array

Suppose each row stores `[student_id, marks]`. The second column is at index 1.

```python
students = np.array([[101, 45],
                     [102, 72],
                     [103, 88],
                     [104, 51]])

passed_students = students[students[:, 1] >= 50]
print(passed_students)
```

## 10. `np.where()`

`np.where(condition)` returns the indices where a condition is true. With three arguments, it chooses values from the second or third argument depending on the condition.

```python
marks = np.array([35, 48, 67, 82, 55])

print(np.where(marks >= 50))  # indices where condition is True
print(np.where(marks >= 50, "Pass", "Fail"))  # conditional values
```

**Distinction:** `arr[condition]` returns matching values. `np.where(condition)` returns their indices. `np.where(condition, x, y)` returns values selected from `x` or `y`.

## 11. Arithmetic operations

Most arithmetic operators work element by element when array shapes are compatible.

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)  # element-wise multiplication
print(a / b)
print(a ** 2)
print(a + 5)  # scalar applied to every element
```

**Do not confuse:** `a * b` is element-wise multiplication. Matrix multiplication is a different operation, usually written `a @ b` for compatible 2D arrays.

## 12. Broadcasting

Broadcasting allows NumPy to perform element-wise operations on arrays with different but compatible shapes, without manually copying values.

Compare dimensions from right to left. Two dimensions are compatible if they are equal, or if one of them is 1. Missing leading dimensions are treated as 1.

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print(arr + 5)  # scalar broadcasts to every element

row = np.array([1, 2, 3])
print(arr + row)  # row is applied to each row
```

### Broadcasting with a column

A column-shaped array `(2, 1)` can combine with a row-shaped array `(3,)`, producing a `(2, 3)` result.

```python
column = np.array([[100],
                   [200]])       # shape (2, 1)
row = np.array([1, 2, 3])         # shape (3,)

print(column + row)
print((column + row).shape)
```

**Shape check:** `(2, 1)` and `(3,)` are treated as `(2, 1)` and `(1, 3)`. The dimensions are compatible, so the result is `(2, 3)`. Shapes `(2, 3)` and `(2, 2)` are not broadcast-compatible.

## 13. Aggregations and statistics

Aggregation reduces many values to a summary value.

```python
marks = np.array([40, 50, 60, 70, 80, 90])

print("sum:", np.sum(marks))
print("mean:", np.mean(marks))
print("median:", np.median(marks))
print("min:", np.min(marks))
print("max:", np.max(marks))
print("range:", np.max(marks) - np.min(marks))
```

### Median

The median is the middle value after sorting. With an even number of values, it is the average of the two middle values.

```python
print(np.median([10, 20, 30]))       # 20
print(np.median([10, 20, 30, 40]))  # 25
```

### Percentiles

A percentile describes a position in a distribution. The 50th percentile is the median. The 25th and 75th percentiles are often called Q1 and Q3.

```python
data = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print(np.percentile(data, 25))
print(np.percentile(data, 50))
print(np.percentile(data, 75))
```

### Variance and standard deviation

Variance measures the average squared distance from the mean. Standard deviation is the square root of variance and is expressed in the same units as the data.

- Low standard deviation: values tend to be closer to the mean.
- High standard deviation: values tend to be more spread out.

By default, NumPy uses the population formula (`ddof=0`). Use `ddof=1` for the common sample estimate.

```python
data = np.array([40, 50, 60, 70, 80, 90])

print("Population variance:", np.var(data))
print("Population std:", np.std(data))
print("Sample variance:", np.var(data, ddof=1))
print("Sample std:", np.std(data, ddof=1))
```

**What does `ddof` mean?** It is the *delta degrees of freedom* adjustment. NumPy divides by `N - ddof` when calculating variance. `ddof=0` divides by `N`; `ddof=1` divides by `N-1`. Use the sample version when treating the data as a sample used to estimate a larger population.

### Statistics along an axis

For a 2D array:
- `axis=0`: reduce down the rows; you get one result per column.
- `axis=1`: reduce across the columns; you get one result per row.

```python
scores = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print("Column sums:", np.sum(scores, axis=0))
print("Row sums:", np.sum(scores, axis=1))
print("Column means:", np.mean(scores, axis=0))
```

### `argmin()` and `argmax()`

`argmin` and `argmax` return the index of the smallest or largest value—not the value itself.

```python
values = np.array([100, 500, 200, 800, 300])

print(np.max(values))     # largest value
print(np.argmax(values))  # index of largest value
print(np.argmin(values))  # index of smallest value
```

## 14. Random number generation

NumPy provides tools to generate random samples for simulations, experiments, and machine learning workflows.

```python
np.random.seed(42)  # makes this legacy random sequence reproducible
print(np.random.randint(1, 10, size=5))  # integers from 1 up to (not including) 10
print(np.random.rand(5))                 # uniform values in [0, 1)
print(np.random.randn(5))                # standard normal samples
```

### Reproducibility with a seed

A fixed seed lets you reproduce a pseudo-random sequence. The number `42` is not mathematically special; any fixed integer can be used.

```python
np.random.seed(7)
first_run = np.random.randint(0, 100, size=5)

np.random.seed(7)
second_run = np.random.randint(0, 100, size=5)

print(first_run)
print(second_run)
print(np.array_equal(first_run, second_run))
```

For new code, NumPy recommends using the newer random Generator API (`default_rng`).

```python
rng = np.random.default_rng(42)
print(rng.integers(1, 10, size=5))
print(rng.random(5))
```

### `choice()` and replacement

`choice()` samples from a collection. Sampling is with replacement by default, so an item may be selected more than once. Use `replace=False` to avoid repeats (and ensure the requested sample size is not larger than the population).

```python
items = np.array(["A", "B", "C", "D"])

rng = np.random.default_rng(1)
print(rng.choice(items, size=5))  # replacement is allowed
print(rng.choice(items, size=3, replace=False))
```

### `shuffle()`

`shuffle()` rearranges the original array in place. It does not return a newly shuffled array.

```python
values = np.array([1, 2, 3, 4, 5])
rng = np.random.default_rng(10)
rng.shuffle(values)
print(values)
```

## 15. Linear algebra fundamentals

Linear algebra is important in machine learning because data, weights, transformations, and many model calculations are represented using vectors and matrices. This section focuses on the NumPy operations and their intuition.

### Vectors and matrices

A vector is a 1D array of values. A matrix is a 2D array arranged in rows and columns. The `shape` tells you the dimensions.

```python
vector = np.array([2, 3])
matrix = np.array([[1, 2],
                   [3, 4]])

print(vector.shape)  # (2,)
print(matrix.shape)  # (2, 2)
```

### Vector addition and scalar multiplication

Vectors of the same shape can be added element by element. A scalar multiplies every vector element.

```python
v1 = np.array([2, 3])
v2 = np.array([4, 5])

print(v1 + v2)
print(3 * v1)
```

### Dot product

For two vectors of equal length, the dot product multiplies corresponding entries and adds the products. It returns a single number.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a * b)          # element-wise products: [4, 10, 18]
print(np.dot(a, b))   # dot product: 32
print(a @ b)          # same dot product for 1D vectors
```

### Matrix multiplication

Use `@` or `np.matmul()` for matrix multiplication. If A has shape `(m, n)` and B has shape `(n, p)`, the result has shape `(m, p)`. The inner dimensions must match.

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print(A @ B)
print((A @ B).shape)
```

### Transpose

Transpose swaps rows and columns. For a 2D array, use `.T`.

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print(matrix.T)
print(matrix.shape, matrix.T.shape)
```

### Identity matrix and inverse

An identity matrix behaves like 1 in matrix multiplication: `A @ I = A` when shapes match. The inverse of a square matrix reverses its transformation when it exists: `A @ inv(A)` is approximately the identity matrix. A matrix with determinant zero is singular and has no inverse.

```python
A = np.array([[4.0, 0.0],
              [0.0, 4.0]])

A_inverse = np.linalg.inv(A)
identity = A @ A_inverse

print("Inverse:")
print(A_inverse)
print("A multiplied by its inverse:")
print(identity)
```

### Determinant

The determinant is a scalar associated with a square matrix. In 2D, its absolute value describes the area-scaling factor of the transformation. If the determinant is zero, the transformation collapses area and the matrix is not invertible.

```python
A = np.array([[4.0, 0.0],
              [0.0, 4.0]])
print(np.linalg.det(A))
```

### Eigenvalues and eigenvectors — intuition

An eigenvector is a non-zero direction that a matrix transformation leaves on the same line; the eigenvalue tells how much that vector is scaled (a negative value also reverses its direction). NumPy returns eigenvectors as columns of the eigenvector matrix.

```python
A = np.array([[2.0, 0.0],
              [0.0, 3.0]])

eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors (columns):")
print(eigenvectors)
```

**Priority note:** For an introductory NumPy / AI-engineering foundation, be comfortable with vector and matrix shapes, dot products, matrix multiplication, transpose, and the purpose of inverse/eigen concepts. You do not need to memorise lengthy hand calculations before progressing to data handling and ML.

## 16. Performance and vectorization

Vectorization means expressing an operation on an entire array instead of writing an explicit Python loop to process each element. NumPy can execute many such operations in optimized compiled code.

```python
numbers = np.array([10, 20, 30, 40])

# Explicit Python loop
loop_result = []
for number in numbers:
    loop_result.append(number * 2)

# Vectorized operation
vectorized_result = numbers * 2

print(loop_result)
print(vectorized_result)
```

### A practical marks example

```python
marks = np.array([40, 50, 60, 70, 80])

# Add five marks to every student's score in one vectorized expression
updated_marks = marks + 5

print(updated_marks)
```

**Interview explanation:** Vectorization applies operations to whole arrays without writing Python-level loops. It often improves performance and makes numerical code shorter and clearer. For fair timing comparisons, use sufficiently large arrays and benchmark carefully; tiny examples may not show a speed advantage.

```python
import time

large = np.arange(1_000_000)

start = time.perf_counter()
loop_result = [value * 2 for value in large]
loop_seconds = time.perf_counter() - start

start = time.perf_counter()
array_result = large * 2
numpy_seconds = time.perf_counter() - start

print(f"Python loop: {loop_seconds:.6f} seconds")
print(f"NumPy vectorized: {numpy_seconds:.6f} seconds")
```

## 17. Common mistakes and interview notes

1. **Confusing `*` and `@`:** `*` is element-wise multiplication; `@` is matrix multiplication.
2. **Forgetting slice stop is excluded:** `arr[1:4]` includes indices 1, 2, and 3.
3. **Using `and` / `or` with arrays:** use `&` / `|` and wrap comparisons in parentheses.
4. **Confusing values and indices:** `max()` returns a value; `argmax()` returns an index.
5. **Assuming `reshape()` can change element count:** the total number of elements must stay the same.
6. **Confusing view and copy:** a view shares memory; a copy is independent.
7. **Misreading axes:** for a 2D array, `axis=0` reduces rows (one result per column); `axis=1` reduces columns (one result per row).
8. **Forgetting broadcasting rules:** dimensions must match or one of them must be 1.
9. **Misunderstanding `ddof`:** it changes the divisor used for variance and standard deviation.
10. **Assuming random output is repeatable:** set/reset a seed or use a seeded Generator when reproducibility is needed.
